```python
weekly_accidents = [3,0,5,8,2,12,1]

for accident in weekly_accidents:
    if accident==0:
       print("Zero incidents record")
    elif accident<=3:
       print("Low incident count")
    elif accident<=5:
       print("Moderate incidents count")
    else:
       print("High incident count")
```
